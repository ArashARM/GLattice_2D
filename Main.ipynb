```python
import numpy as np
from Config import Config
from Domain import Domain
from GLattice2D import GLattice2D
from LatticeCell2D import LatticeCell2D
from LatticeStructure2D import LatticeStructure2D
```

```python
config = Config(
    M=1,
    step_size=2,
    passage_angle=80.0,
    required_weight=0.001,
    strut_width=0.03,
    random_seed=None,
    max_direction_attempts= 200,
    max_particle_attempts = 50,
    max_generation_attempts= 20,
)

#config.show()
# --------------------------------
# 2. Create square design domain
# --------------------------------

domain = Domain(
    config.M
)

# --------------------------------
# 3. Create lattice generator
# --------------------------------

lattice = GLattice2D(
    domain,
    config
)
lattice.generate()
lattice.plot()
```

```python
cell = LatticeCell2D(
    lattice
)

cell.build()
cell.plot()
```

```python
structure = LatticeStructure2D(
    cell,
    n_x=4,
    n_y=3
)

structure.build()
structure.plot()
```
